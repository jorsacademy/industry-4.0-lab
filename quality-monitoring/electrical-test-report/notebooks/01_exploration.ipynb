# Electrical Test Report — exploration
Lot-aware smart-quality exploration for the Kaggle electrical test dataset.

````python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from src.data import load_electrical_report, measurement_columns
from src.monitoring import summarize_lots
````

Download first if the raw CSV is not present:
```bash
python scripts/download_data.py
```

````python
path = Path("data/raw/Electrical_Test_Report.csv")
df = load_electrical_report(path)
df.shape, df["is_defect"].mean(), df["LOT"].nunique(), measurement_columns(df)
````

````python
df["is_defect"].value_counts().sort_index().plot.bar()
plt.title("Electrical test class balance")
plt.xlabel("Defect flag")
plt.ylabel("Records")
plt.show()
````

````python
lots = summarize_lots(df, min_lot_size=100, warning_rate=0.05)
lots[lots["stable_lot"]]["defect_rate"].plot.hist(bins=30)
plt.axvline(0.05, linestyle="--")
plt.title("Stable-lot defect-rate distribution")
plt.xlabel("Defect rate")
plt.show()
````
