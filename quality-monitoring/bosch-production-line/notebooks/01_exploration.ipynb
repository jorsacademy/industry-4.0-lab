# Bosch production-line exploration
This notebook inspects generated station-level features. Raw competition data stays outside Git.

```python
from pathlib import Path
import pandas as pd

path = Path('../data/processed/bosch_features_p100.parquet')
features = pd.read_parquet(path)
features.shape
```

```python
features['Response'].value_counts(normalize=True).sort_index()
```

```python
features[['process_duration', 'prefix_station_count', 'numeric_missing_ratio']].describe(percentiles=[.01,.1,.5,.9,.99])
```

```python
station_cols = [c for c in features if c.endswith('__seen')]
features.groupby('Response')[station_cols].mean().T.assign(delta=lambda x: x[1]-x[0]).sort_values('delta', key=abs, ascending=False).head(20)
```
