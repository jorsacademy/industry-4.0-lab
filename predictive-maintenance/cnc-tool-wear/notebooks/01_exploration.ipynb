# CNC Tool Wear — Sensor Exploration

This notebook inspects run-level labels and multi-axis CNC telemetry before model training.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve()
RAW = ROOT / 'data' / 'raw'
metadata = pd.read_csv(RAW / 'train.csv')
metadata
```

```python
metadata['tool_condition'].value_counts().plot(kind='bar', title='Tool condition by experiment')
plt.ylabel('experiments')
plt.show()
```

```python
run = pd.read_csv(RAW / 'experiment_01.csv')
print(run.shape)
run.head()
```

```python
candidate = [c for c in ['S1_OutputPower', 'S1_CurrentFeedback', 'S1_ActualVelocity', 'M1_CURRENT_FEEDRATE'] if c in run.columns]
run[candidate].plot(figsize=(14, 7), subplots=True, sharex=True, title='Experiment 01 controller telemetry')
plt.xlabel('sample (10 Hz)')
plt.tight_layout()
plt.show()
```

```python
process_counts = run['Machining_Process'].value_counts()
process_counts
```

## Leakage note
Rows from one experiment share one tool-condition label. Do not perform a random row split. The training pipeline groups all windows by experiment id.
